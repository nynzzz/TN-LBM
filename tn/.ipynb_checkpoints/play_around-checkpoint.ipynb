# Tensor Network Exploration with Quimb

Learning the basics of MPS (Matrix Product State) and QTT (Quantics Tensor Train) representations for LBM.

```python
import numpy as np
import matplotlib.pyplot as plt
import quimb as qu
import quimb.tensor as qtn

print(f"quimb version: {qu.__version__}")
```

## 1. MPS Basics

An MPS (Matrix Product State) represents a 1D tensor as a chain of smaller tensors.
For a state with L sites and physical dimension d, we have:

$$\Psi_{i_1 i_2 \ldots i_L} = \sum_{\alpha} A^{(1)}_{i_1,\alpha_1} A^{(2)}_{\alpha_1,i_2,\alpha_2} \ldots A^{(L)}_{\alpha_{L-1},i_L}$$

```python
# Create a random MPS with 8 sites, physical dimension 2, bond dimension 4
L = 8  # number of sites
d = 2  # physical dimension (like qubit)
chi = 4  # bond dimension

psi = qtn.MPS_rand_state(L=L, bond_dim=chi, phys_dim=d)

print(f"Number of sites: {psi.L}")
print(f"Physical dimension: {psi.phys_dim()}")
print(f"Max bond dimension: {psi.max_bond()}")
print(f"Number of parameters: {sum(t.size for t in psi)}")
print(f"Full tensor would have: {d**L} parameters")
```

```python
# Visualize the MPS structure
psi.draw(color=['I0', 'I1', 'I2', 'I3', 'I4', 'I5', 'I6', 'I7'])
```

```python
# Look at individual tensors
print("Tensor shapes in the MPS:")
for i, t in enumerate(psi):
    print(f"  Site {i}: shape {t.shape}, indices {t.inds}")
```

## 2. MPS from a Vector (QTT Encoding)

For our LBM application, we want to encode a 1D field (like density) as an MPS.
With N = 2^L grid points, we use QTT (Quantics TT) encoding where each site has physical dimension 2.

```python
# Create a simple 1D field (e.g., a Gaussian)
L = 8  # number of qubits -> 2^8 = 256 grid points
N = 2**L
x = np.linspace(0, 1, N)
sigma = 0.1
field = np.exp(-((x - 0.5)**2) / (2 * sigma**2))

plt.figure(figsize=(10, 3))
plt.plot(x, field)
plt.xlabel('x')
plt.ylabel('field value')
plt.title(f'Original field (N={N} points)')
plt.show()

print(f"Field shape: {field.shape}")
print(f"Field norm: {np.linalg.norm(field):.6f}")
```

```python
# Reshape field for QTT encoding: (N,) -> (2, 2, 2, ..., 2) with L dimensions
# This is the dyadic mapping: x = sum_k q_k * 2^(L-k)
field_tensor = field.reshape([2] * L)
print(f"Reshaped tensor shape: {field_tensor.shape}")
```

```python
# Convert to MPS using quimb's tensor_1d_to_mps function
# This performs SVD decomposition to create an MPS
mps = qtn.MatrixProductState.from_dense(field_tensor, dims=[2]*L)

print(f"MPS sites: {mps.L}")
print(f"Max bond dimension: {mps.max_bond()}")
print(f"Bond dimensions: {[mps.bond_size(i, i+1) for i in range(L-1)]}")
```

```python
# Reconstruct the field from MPS and compare
reconstructed_tensor = mps.to_dense()
reconstructed_field = reconstructed_tensor.flatten()

error = np.linalg.norm(field - reconstructed_field) / np.linalg.norm(field)
print(f"Relative reconstruction error: {error:.2e}")

plt.figure(figsize=(10, 3))
plt.plot(x, field, 'b-', label='Original', linewidth=2)
plt.plot(x, reconstructed_field, 'r--', label='Reconstructed', linewidth=1)
plt.xlabel('x')
plt.ylabel('field value')
plt.legend()
plt.title('MPS reconstruction (no truncation)')
plt.show()
```

## 3. MPS Compression (Truncation)

The power of MPS comes from compression: we can truncate small singular values to reduce bond dimension while maintaining accuracy.

```python
# Test different bond dimension limits
chi_values = [2, 4, 8, 16, 32]

plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(x, field, 'k-', label='Original', linewidth=2)

errors = []
for chi_max in chi_values:
    # Create MPS with truncation
    mps_trunc = qtn.MatrixProductState.from_dense(
        field_tensor, dims=[2]*L, max_bond=chi_max
    )
    recon = mps_trunc.to_dense().flatten()
    err = np.linalg.norm(field - recon) / np.linalg.norm(field)
    errors.append(err)
    
    plt.plot(x, recon, '--', label=f'chi={chi_max}, err={err:.1e}')

plt.xlabel('x')
plt.ylabel('field')
plt.legend(fontsize=8)
plt.title('Compression at different bond dimensions')

plt.subplot(1, 2, 2)
plt.semilogy(chi_values, errors, 'bo-')
plt.xlabel('Max bond dimension')
plt.ylabel('Relative error')
plt.title('Error vs bond dimension')
plt.grid(True)

plt.tight_layout()
plt.show()
```

## 4. MPS Arithmetic

For LBM, we need:
- Addition: $|\rho\rangle = \sum_i |f_i\rangle$
- Scalar multiplication: $\omega |f\rangle$
- Hadamard (element-wise) product: $|f\rangle \circ |g\rangle$ (for collision)

```python
# Create two test fields
field1 = np.exp(-((x - 0.3)**2) / (2 * 0.05**2))
field2 = np.exp(-((x - 0.7)**2) / (2 * 0.05**2))

# Convert to MPS
mps1 = qtn.MatrixProductState.from_dense(field1.reshape([2]*L), dims=[2]*L)
mps2 = qtn.MatrixProductState.from_dense(field2.reshape([2]*L), dims=[2]*L)

print(f"MPS1 max bond: {mps1.max_bond()}")
print(f"MPS2 max bond: {mps2.max_bond()}")
```

```python
# Addition: MPS1 + MPS2
# In quimb, we can add MPS directly (bond dimensions add)
mps_sum = mps1.add_MPS(mps2)

print(f"Sum MPS max bond: {mps_sum.max_bond()} (was {mps1.max_bond()} + {mps2.max_bond()})")

# Verify
sum_recon = mps_sum.to_dense().flatten()
sum_exact = field1 + field2
print(f"Addition error: {np.linalg.norm(sum_recon - sum_exact) / np.linalg.norm(sum_exact):.2e}")

plt.figure(figsize=(10, 3))
plt.plot(x, sum_exact, 'b-', label='Exact sum', linewidth=2)
plt.plot(x, sum_recon, 'r--', label='MPS sum')
plt.legend()
plt.title('MPS Addition')
plt.show()
```

```python
# Compress after addition to control bond dimension
mps_sum_compressed = mps_sum.compress(max_bond=16)
print(f"Compressed sum MPS max bond: {mps_sum_compressed.max_bond()}")

sum_compressed = mps_sum_compressed.to_dense().flatten()
print(f"Compression error: {np.linalg.norm(sum_compressed - sum_exact) / np.linalg.norm(sum_exact):.2e}")
```

```python
# Scalar multiplication
alpha = 2.5
mps_scaled = mps1.multiply(alpha, inplace=False)

scaled_recon = mps_scaled.to_dense().flatten()
scaled_exact = alpha * field1
print(f"Scalar mult error: {np.linalg.norm(scaled_recon - scaled_exact) / np.linalg.norm(scaled_exact):.2e}")
```

## 5. Hadamard (Element-wise) Product

This is the critical operation for the collision step. The naive implementation squares the bond dimension!

```python
# Element-wise product using quimb
# We need to compute psi1 * psi2 element-wise

# Method: Contract both MPS with a "copying" tensor that duplicates indices
# For now, let's use a direct approach via dense computation

hadamard_exact = field1 * field2

# Convert Hadamard product to MPS
mps_hadamard = qtn.MatrixProductState.from_dense(
    hadamard_exact.reshape([2]*L), dims=[2]*L
)

print(f"Hadamard MPS max bond: {mps_hadamard.max_bond()}")

plt.figure(figsize=(10, 3))
plt.plot(x, field1, 'b-', alpha=0.5, label='field1')
plt.plot(x, field2, 'g-', alpha=0.5, label='field2')
plt.plot(x, hadamard_exact, 'r-', linewidth=2, label='field1 * field2')
plt.legend()
plt.title('Hadamard product')
plt.show()
```

## 6. MPO (Matrix Product Operator)

For streaming, we need shift operators. An MPO acts on an MPS to produce another MPS.

```python
# Create a simple identity MPO
mpo_identity = qtn.MPO_identity(L=L, phys_dim=2)

print(f"Identity MPO sites: {mpo_identity.L}")
print(f"MPO max bond: {mpo_identity.max_bond()}")

# Apply MPO to MPS
result = mpo_identity.apply(mps1)
result_dense = result.to_dense().flatten()

print(f"Identity application error: {np.linalg.norm(result_dense - field1):.2e}")
```

## 7. Next Steps for LBM

To implement LBM in tensor network format, we need:

1. **QTT encoding**: Convert 1D fields to MPS (done above)
2. **Shift MPO**: For streaming, construct MPO that shifts indices by 1
3. **Efficient Hadamard**: Implement TTM or HaTT algorithm for collision
4. **Mask operations**: Binary masks for boundary conditions

Key challenges:
- Streaming: How to efficiently represent the shift operator as an MPO?
- Collision: How to keep bond dimension bounded during repeated Hadamard products?

```python
# Quick test: memory scaling
L_values = [6, 8, 10, 12, 14, 16]
chi_test = 8

print("Memory comparison: full tensor vs MPS")
print(f"{'L':>4} {'N':>10} {'Full (KB)':>12} {'MPS (KB)':>12} {'Ratio':>10}")
print("-" * 50)

for L in L_values:
    N = 2**L
    full_mem = N * 8 / 1024  # 8 bytes per float64
    # MPS memory: roughly L * chi^2 * d * 8 bytes
    mps_mem = L * chi_test**2 * 2 * 8 / 1024
    ratio = full_mem / mps_mem
    print(f"{L:>4} {N:>10} {full_mem:>12.2f} {mps_mem:>12.2f} {ratio:>10.1f}x")
```
